# Analyze the results

## Load the data

```python
from pandas import read_csv
import matplotlib.pyplot as plt
load_python = False

if load_python:
    numerical_columns = ['AND', 'OR', 'Total Conditions', 'Max Depth', 'Max Repeated']
    result_name = 'results-python.csv'
    result = read_csv(result_name)
    print(result.head(5))
else:
    numerical_columns = ['AND', 'OR', 'Total Conditions']
    result_name = 'results-cpp.csv'
    result_with_one_condition = read_csv(result_name)
    result = result_with_one_condition[result_with_one_condition['Total Conditions'] > 1]
    print(result.head(5))
```

## Helper Functions

```python
def get_nummerical_results(df):
    for column in numerical_columns:
        print(f"Statistics for {column}:")
        print(f"Average: {df[column].mean()}")
        print(f"Minimum: {df[column].min()}")
        print(f"Maximum: {df[column].max()}")
        print("-" * 40)

def show_frequency(df, column):
    plt.figure(figsize=(10, 6))  # Size of the figure
    df[column].hist(bins=50, edgecolor='black')  # Use bins=50 or adjust as necessary
    plt.title(f'Histogram for {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.grid(False)  # Remove gridlines
    plt.show()


def get_frequency_of_nummerical(df):
    # Plot histograms for each numerical column
    for column in numerical_columns:
        show_frequency(df, column)

def get_rows_with_condition_type(df, condition_type):
    return df[df['Condition Type'] == condition_type]

def get_rows_within_directory(df, directory):
    return df[df['File'].str.contains(directory)]
```

## Nummerical results

```python
get_nummerical_results(result)
```

```python
get_frequency_of_nummerical(result)
```

```python
show_frequency(result, 'Decision Type')
```
